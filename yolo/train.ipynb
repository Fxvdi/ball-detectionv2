```python
from ultralytics import YOLO

# 1. Modellwahl: 'x' ist sehr mächtig, aber bei 100 Bildern riskant (Overfitting)
# Falls es zu langsam ist oder schlecht generalisiert, versuch mal 'yolo11m.pt'
model = YOLO("yolo26m.pt") # Ich nehme an, du meinst die aktuellste Version

# 2. Finetuning mit Anpassungen
model.train(
    data="../data.yaml",
    epochs=100,
    imgsz=640,        # Top für kleine Objekte wie Handbälle
    batch=8,          # Okay, solange dein VRAM (GPU-Speicher) mitspielt
    device=0,
    workers=4,
    pretrained=True,
    # --- Zusätzliche Tipps ---
    patience=20,      # Stoppt das Training, wenn sich 20 Epochen lang nichts verbessert
    save=True,        # Speichert Checkpoints
    augment=True,     # WICHTIG bei nur 100 Bildern (aktiviert automatische Albumentations)
    project="Handball_Detection", 
    name="v1_extra_large"
)
```
