```python
from ultralytics import YOLO
```

```python
model = YOLO("yolo26x.pt")
data_yaml = "../train/data.yaml"
```

```python
results = model.train(
    data=data_yaml,
    epochs=50,
    imgsz=512,
    batch=8,
    patience=5,
    device=0,
    project='runs/detect',
    name='test_small_416',
    plots=True,
    save=True,
    verbose=True
)
```
